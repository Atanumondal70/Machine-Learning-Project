```python
# CELL 1 — IMPORT LIBRARIES
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report, roc_auc_score,
    roc_curve, ConfusionMatrixDisplay
)
from sklearn.inspection import permutation_importance
import joblib

sns.set_theme(style="whitegrid", palette="Set2")
RANDOM_STATE = 42
print("Libraries imported successfully.")
```

```python
# CELL 2 — LOAD DATASET
DATA_PATH = Path("organic_farming_suitability_1000_source_calibrated.csv")

if not DATA_PATH.exists():
    # Helpful fallback if the notebook is run from another folder
    DATA_PATH = Path("C://Users//ashiq//Downloads//organic_farming_suitability_1000_source_calibrated.csv")

df = pd.read_csv(DATA_PATH)
print(f"Dataset loaded: {df.shape[0]} rows × {df.shape[1]} columns")
display(df.head())
```

```python
# CELL 3 — DATASET OVERVIEW
print("Shape:", df.shape)
print("\\nColumn names:")
print(df.columns.tolist())
print("\\nData types:")
display(df.dtypes.to_frame("Data_Type"))
print("\\nDataset information:")
df.info()
```

```python
# CELL 4 — STATISTICAL SUMMARY
display(df.describe(include="all").T)
```

```python
# CELL 5 — DATA QUALITY CHECKS
df.columns = df.columns.str.strip()
print("Missing values per column:")
display(df.isna().sum().to_frame("Missing_Count"))
print("\\nDuplicate complete rows:", df.duplicated().sum())
if "Sample_ID" in df.columns:
    print("Duplicate Sample_ID values:", df["Sample_ID"].duplicated().sum())

numeric_cols_check = df.select_dtypes(include=np.number).columns
print("Infinite numeric values:", np.isinf(df[numeric_cols_check]).sum().sum())
```

```python
# CELL 6 — CLEANING
df = df.replace([np.inf, -np.inf], np.nan).drop_duplicates().copy()
if "Sample_ID" in df.columns:
    df = df.drop_duplicates(subset="Sample_ID", keep="first")

TARGET = "Organic_Farming_Suitability"
assert TARGET in df.columns, f"Target column {TARGET!r} was not found."
df = df.dropna(subset=[TARGET]).copy()
print("Cleaned shape:", df.shape)
print("Target values:", df[TARGET].unique())
display(df[TARGET].value_counts().to_frame("Count"))
```

```python
# CELL 7 — TARGET DISTRIBUTION
plt.figure(figsize=(7, 4))
ax = sns.countplot(data=df, x=TARGET, order=df[TARGET].value_counts().index)
ax.set_title("Organic Farming Suitability — Target Distribution")
ax.set_xlabel("Class")
ax.set_ylabel("Number of records")
for container in ax.containers:
    ax.bar_label(container)
plt.tight_layout()
plt.axis(False)
plt.show()

display((df[TARGET].value_counts(normalize=True) * 100).round(2).to_frame("Percentage"))
```

```python
# CELL 8 — NUMERIC FEATURE DISTRIBUTIONS
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
numeric_cols = [c for c in numeric_cols if c not in ["Suitability_Score"]]
if numeric_cols:
    df[numeric_cols].hist(figsize=(16, 12), bins=25, edgecolor="white")
    plt.suptitle("Numeric Feature Distributions", y=1.02, fontsize=16)
    plt.tight_layout()
    plt.axis(False)
    plt.show()
```

```python
# CELL 9 — CATEGORICAL FEATURE COUNTS
categorical_cols = df.select_dtypes(include=["object", "category"]).columns.tolist()
categorical_cols = [c for c in categorical_cols if c not in [TARGET, "Sample_ID"]]
for col in categorical_cols:
    plt.figure(figsize=(8, 3.5))
    order = df[col].value_counts().index
    sns.countplot(data=df, y=col, order=order)
    plt.title(f"Distribution: {col}")
    plt.tight_layout()
    plt.axis(False)
    plt.show()
```

```python
# CELL 10 — CORRELATION HEATMAP
corr_cols = df.select_dtypes(include=np.number).columns.tolist()
corr_cols = [c for c in corr_cols if c != "Suitability_Score"]
if len(corr_cols) >= 2:
    plt.figure(figsize=(13, 9))
    sns.heatmap(df[corr_cols].corr(), cmap="coolwarm", center=0, annot=False)
    plt.title("Correlation Heatmap (Numeric Predictors)")
    plt.tight_layout()
    plt.axis(False)
    plt.show()
```

```python
# CELL 11 — DEFINE X AND y
drop_from_X = [
    TARGET,
    "Sample_ID", 
    "Suitability_Score"
]
X = df.drop(
    columns=[c for c in drop_from_X if c in df.columns]).copy()
y = df[TARGET].astype(str).str.strip()

# Standardize label spelling and map to binary 0/1 for ROC-AUC
label_map = {
    "Not Suitable": 0,
    "Suitable": 1
}
unknown_labels = set(y.unique()) - set(label_map)
if unknown_labels:
    raise ValueError(
        f"Unexpected target labels: {
    unknown_labels
    }. Please inspect the dataset."
                    )
y_binary = y.map(label_map)

numeric_features = X.select_dtypes(include=np.number).columns.tolist()
categorical_features = X.select_dtypes(
    include=[
        "object", 
        "category", 
        "bool"]
).columns.tolist()
print("Predictor count:", X.shape[1])
print("Numeric features:", numeric_features)
print("Categorical features:", categorical_features)
print("Target mapping:", label_map)
```

```python
# CELL 12 — TRAIN/TEST SPLIT
X_train, X_test, y_train, y_test, y_train_binary, y_test_binary = train_test_split(
    X, y, y_binary,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)
print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])
print("\\nTraining target distribution:")
display(y_train.value_counts(normalize=True).round(3).to_frame("Proportion"))
```

```python
# CELL 13 — PREPROCESSING PIPELINES
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])
print("Preprocessing pipeline ready.")
```

```python
# CELL 14 — DEFINE CLASSIFIERS
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000, 
        class_weight=None, 
        random_state=RANDOM_STATE
    ),
    "Decision Tree": DecisionTreeClassifier(random_state=RANDOM_STATE, class_weight=None),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "Random Forest": RandomForestClassifier(
        n_estimators=300, random_state=RANDOM_STATE, n_jobs=-1, class_weight=None
    )
}
pipelines = {
    name: Pipeline(steps=[("preprocessor", preprocessor), ("classifier", estimator)])
    for name, estimator in models.items()
}
print("Models:", list(pipelines.keys()))
```

```python
# CELL 15 — TRAIN AND EVALUATE MODELS
results = []
fitted_models = {}

for name, pipe in pipelines.items():
    pipe.fit(X_train, y_train)
    fitted_models[name] = pipe
    pred = pipe.predict(X_test)

    # Probability for the positive class "Suitable"
    class_list = list(pipe.named_steps["classifier"].classes_)
    suitable_index = class_list.index("Suitable")
    proba = pipe.predict_proba(X_test)[:, suitable_index]
    y_true_binary = (y_test == "Suitable").astype(int)

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred, pos_label="Suitable", zero_division=0),
        "Recall": recall_score(y_test, pred, pos_label="Suitable", zero_division=0),
        "F1_Score": f1_score(y_test, pred, pos_label="Suitable", zero_division=0),
        "ROC_AUC": roc_auc_score(y_true_binary, proba)
    })

results_df = pd.DataFrame(results).sort_values("F1_Score", ascending=False).reset_index(drop=True)
display(results_df.round(4))
```

```python
# CELL 16 — MODEL COMPARISON CHART
metric_cols = [
    "Accuracy", 
    "Precision", 
    "Recall", 
    "F1_Score", 
    "ROC_AUC"]
results_long = results_df.melt(id_vars="Model", value_vars=metric_cols,
                               var_name="Metric", value_name="Score")
plt.figure(figsize=(12, 5))
sns.barplot(data=results_long, x="Model", y="Score", hue="Metric")
plt.ylim(0, 1.05)
plt.title("Classifier Performance Comparison (Test Set)")
plt.xticks(rotation=15)
plt.tight_layout()
plt.axis(False)
plt.show()
```

```python
# CELL 17 — BEST MODEL SELECTION
# Choose the model with highest test F1; use ROC-AUC as a tie-breaker.
best_model_name = results_df.sort_values(["F1_Score", "ROC_AUC"], ascending=False).iloc[0]["Model"]
best_model = fitted_models[best_model_name]
print("Selected model:", best_model_name)
print("Selection rule: highest test F1-score, ROC-AUC tie-breaker.")
```

```python
# CELL 18 — DETAILED EVALUATION OF BEST MODEL
best_pred = best_model.predict(X_test)
best_classes = list(best_model.named_steps["classifier"].classes_)
suitable_idx = best_classes.index("Suitable")
best_proba = best_model.predict_proba(X_test)[:, suitable_idx]
y_test_bin = (y_test == "Suitable").astype(int)

print(classification_report(y_test, best_pred, labels=["Not Suitable", "Suitable"], zero_division=0))
cm = confusion_matrix(y_test, best_pred, labels=["Not Suitable", "Suitable"])
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["Not Suitable", "Suitable"])
disp.plot(cmap="Blues", values_format="d")
plt.title(f"Confusion Matrix — {best_model_name}")
plt.tight_layout()
plt.axis(False)
plt.show()
```

```python
# CELL 19 — ROC CURVES FOR ALL MODELS
plt.figure(figsize=(8, 6))
for name, pipe in fitted_models.items():
    classes = list(pipe.named_steps["classifier"].classes_)
    pos_idx = classes.index("Suitable")
    prob = pipe.predict_proba(X_test)[:, pos_idx]
    fpr, tpr, _ = roc_curve(y_test_bin, prob)
    auc_value = roc_auc_score(y_test_bin, prob)
    plt.plot(fpr, tpr, label=f"{name} (AUC={auc_value:.3f})")
plt.plot([0, 1], [0, 1], "k--", label="Random classifier")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve — Test Set")
plt.legend()
plt.tight_layout()
plt.axis(False)
plt.show()
```

```python

# ============================================================
# CELL 20: PERMUTATION FEATURE IMPORTANCE
# ============================================================

from sklearn.inspection import permutation_importance
from sklearn.metrics import make_scorer, f1_score

# ------------------------------------------------------------
# STEP 1: CREATE CUSTOM F1 SCORER
# ------------------------------------------------------------

# Our target labels are:
# "Not Suitable"
# "Suitable"
#
# We want to calculate F1-score specifically for the
# positive class "Suitable".

suitable_f1_scorer = make_scorer(
    f1_score,
    pos_label="Suitable",
    zero_division=0
)

print("Custom F1 scorer created successfully.")


# ------------------------------------------------------------
# STEP 2: CALCULATE PERMUTATION IMPORTANCE
# ------------------------------------------------------------

perm = permutation_importance(
    estimator=best_model,
    X=X_test,
    y=y_test,
    scoring=suitable_f1_scorer,
    n_repeats=10,
    random_state=RANDOM_STATE,
    n_jobs=-1
)

print("Permutation importance calculated successfully.")


# ------------------------------------------------------------
# STEP 3: CREATE FEATURE IMPORTANCE DATAFRAME
# ------------------------------------------------------------

importance_df = pd.DataFrame({
    "Feature": X_test.columns,
    "Importance_Mean": perm.importances_mean,
    "Importance_Std": perm.importances_std
})

# Sort features from highest importance to lowest

importance_df = importance_df.sort_values(
    by="Importance_Mean",
    ascending=False
).reset_index(drop=True)


# ------------------------------------------------------------
# STEP 4: DISPLAY TOP 15 FEATURES
# ------------------------------------------------------------

print("\nTOP 15 IMPORTANT FEATURES")
print("=" * 60)

display(
    importance_df.head(15).round(5)
)


# ------------------------------------------------------------
# STEP 5: VISUALIZE FEATURE IMPORTANCE
# ------------------------------------------------------------

top_imp = (
    importance_df
    .head(15)
    .sort_values(
        by="Importance_Mean",
        ascending=True
    )
)

plt.figure(figsize=(12, 8))

plt.barh(
    top_imp["Feature"],
    top_imp["Importance_Mean"],
    xerr=top_imp["Importance_Std"],
    capsize=4
)

plt.xlabel("Decrease in F1-Score After Feature Permutation")

plt.ylabel("Feature")

plt.title(
    "Top 15 Permutation Feature Importance\n"
    "Organic Farming Suitability Prediction",
    fontsize=14,
    fontweight="bold"
)

plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.4
)

plt.tight_layout()
plt.axis(False)
plt.show()
```

```python
# CELL 21 — SAVE MODEL AND RESULTS
OUTPUT_DIR = Path("project_outputs")
OUTPUT_DIR.mkdir(exist_ok=True)

model_path = OUTPUT_DIR / "best_organic_farming_model.pkl"
joblib.dump(best_model, model_path)
results_df.to_csv(OUTPUT_DIR / "model_comparison.csv", index=False)
importance_df.to_csv(OUTPUT_DIR / "permutation_feature_importance.csv", index=False)

print("Saved model:", model_path.resolve())
print("Saved comparison:", (OUTPUT_DIR / "model_comparison.csv").resolve())
print("Saved feature importance:", (OUTPUT_DIR / "permutation_feature_importance.csv").resolve())
```

```python
# CELL 22 — REUSABLE PREDICTION FUNCTION
def predict_organic_suitability(input_data, model_path="project_outputs/best_organic_farming_model.pkl"):
    '''
    input_data: pandas DataFrame with the same predictor column names as X.
    Returns predicted class and probability of Suitable.
    '''
    loaded_model = joblib.load(model_path)
    required_columns = list(X.columns)
    missing = [col for col in required_columns if col not in input_data.columns]
    if missing:
        raise ValueError(f"Input is missing required predictor columns: {missing}")
    prediction = loaded_model.predict(input_data[required_columns])
    classes = list(loaded_model.named_steps["classifier"].classes_)
    suitable_idx = classes.index("Suitable")
    probability = loaded_model.predict_proba(input_data[required_columns])[:, suitable_idx]
    return pd.DataFrame({
        "Predicted_Suitability": prediction,
        "Probability_Suitable": probability
    })

# Example: predict for one row from the test set
example_input = X_test.iloc[[0]].copy()
display(predict_organic_suitability(example_input))
```
